```python
#Q6 Most frequent value
#med = 254

# Médiane initiale (les valeurs manquantes sont ignorées)
median_before = df['horsepower'].median()
print("Médiane avant :", median_before)

# 1. Valeur la plus fréquente
most_frequent = df['horsepower'].mode()[0]
print("Valeur la plus fréquente :", most_frequent)

# 2. Remplir les valeurs manquantes avec cette valeur
df['horsepower'] = df['horsepower'].fillna(most_frequent)

# 3. Nouvelle médiane
median_after = df['horsepower'].median()
print("Médiane après :", median_after)
```

```python
import numpy as np

# 1. Voitures d'Asie
df_asia = df[df['origin'] == 'Asia']

# 2. Colonnes vehicle_weight et model_year
df_asia = df_asia[['vehicle_weight', 'model_year']]

# 3. Les 7 premières lignes
df_asia = df_asia.head(7)

# 4. Tableau NumPy
X = df_asia.values

# 5. Produit X^T X
XTX = X.T.dot(X)

# 6. Inverse
XTX_inv = np.linalg.inv(XTX)

# 7. Vecteur y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. w = (X^T X)^-1 X^T y
w = XTX_inv.dot(X.T).dot(y)
print("w =", w)

# 9. Somme des éléments
print("Somme :", w.sum())
```
